# Problem

I encountered this problem that if you want to make a species somewhere in the mid of chain as focal one you have to assign colours in two directions.

Let's say the species are ordered like this:

```
X  Y  Z  W
```
and Y is focal: 

```
X < Y > Z > W
```

Colours start from chromosome of focal Chr Y must move in two directions:

- from Y to Z, then from Z to W
- from Y to X

This problem is solved with `assign_colors()`. It only prepares the empty colour memory and runs these two directions.

## Read the data

````python
import pandas as pd
import functions

xy = pd.read_csv("tests/data/ac1a.tsv", sep="\t")
yz = pd.read_csv("tests/data/ac1b.tsv", sep="\t")
zw = pd.read_csv("tests/data/ac1c.tsv", sep="\t")
````

````python
print("X-Y")
print(xy)
print("Y-Z")
print(yz)
print("Z-W")
print(zw)
````

## Add colours with the engine function

````python
chain = ["X", "Y", "Z", "W"]
links = {("X", "Y"): xy, ("Y", "Z"): yz, ("Z", "W"): zw}
focal_colors = {"Y1": "#ff0000", "Y2": "#0000ff"}

functions.assign_colors(chain, "Y", focal_colors, links, "#e3e3e3")
````

````python
print("X-Y")
print(xy)
print("Y-Z")
print(yz)
print("Z-W")
print(zw)
````
